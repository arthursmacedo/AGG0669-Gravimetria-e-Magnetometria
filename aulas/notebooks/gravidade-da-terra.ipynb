## Bem-vindos, ao notebook da aula sobre a gravidade da Terra

Nessa aula vamos usar os dados do International Centre for Global Earth Models (ICGEM) para calcular a aceleração da gravidade no planeta.

```python
#Primeiro importamos os dados que baixamos no site do ICGEM
malha = hm.load_icgem_gdf(
    fname="gravity_ell_GOCO06s_cf902582486e3c85f5c2d5582f0a853996f51783f517de0a91f8a65894723f61.gdf"
)
```

```python
#Vamos ver como é a "cara" da nossa malha de dados. Aqui podemos ver as 
# coordenadas, as variáveis e também os atributos da nossa malha
malha
```

```python
#Vamos usar o pyGMT para fazer uma figura da nossa malha
fig = pygmt.Figure()
#Plotando os dados da malha
fig.grdimage(
    malha.gravity_ell, 
    projection="H20c",
    cmap="viridis",
)
#Plotando a linha de costa e os "ticks" do mapa
fig.coast(shorelines=True, frame=True)
#Plotando uma barra de cores para nosso mapa
fig.colorbar()
#Mostrando a figura, o argumento width aumenta o tamanho de exibição da imagem
fig.show(width=1000)
```
